```python
import pandas as pd

# ===============================
# 1) خواندن فایل ورودی
# ===============================
input_path = "1 (6).csv"  #م
df = pd.read_csv(input_path)

print("تعداد سطرهای اولیه:", len(df))

# ===============================
# 2) شمارش نمونه‌هایی که در هر دو ستون 'SMILES' و 'Adduct' تکراری دارند
# ===============================
# شناسایی نمونه‌هایی که در هر دو ستون تکراری دارند
duplicates = df[df.duplicated(subset=['Smiles', 'Adduct'], keep=False)]

# چاپ تعداد و نمایش نمونه‌های تکراری
print(f"تعداد نمونه‌های تکراری (در هر دو ستون SMILES و Adduct): {len(duplicates)}")

# نمایش 10 نمونه اول از نمونه‌های تکراری (قبل از حذف)
print("10 نمونه اول از نمونه‌های تکراری (قبل از حذف):")
print(duplicates.head(10))

# ===============================
# 3) حذف نمونه‌های تکراری
# ===============================
df_cleaned = df.drop_duplicates(subset=['Smiles', 'Adduct'], keep='first').reset_index(drop=True)

print("تعداد سطرها بعد از حذف نمونه‌های تکراری:", len(df_cleaned))

# ===============================
# 4) چاپ 10 ردیف اول از دیتافریم بعد از حذف نمونه‌های تکراری
# ===============================
print("10 ردیف اول دیتافریم بعد از حذف نمونه‌های تکراری:")
print(df_cleaned.head(10))

# ===============================
# 5) ذخیره فایل خروجی
# ===============================
output_path = "output_no_missing_values1.csv"  # مسیر ذخیره فایل خروجی
df_cleaned.to_csv(output_path, index=False)

print(f"فایل ذخیره شد بدون نمونه‌های تکراری → {output_path}")
```
